```python
import sys

def ask_fahrenheit_input_float():                                           # Gets the temperature input and converts to float using a loop.
    mistake = 0                                                             # Tracks the errors across loop iterations
    max_mistakes = 5

    while True:
        try:
            fahrenheit = float(input("What is the temperature in Fahrenheit? "))
            return fahrenheit
        except ValueError:
            mistake += 1
            if mistake == max_mistakes:
                print("Too many mistakes. Terminating the program.")        # Stops the entire program cleanly
                sys.exit()
            elif mistake >= 3:
                print("Wag makulit, sabi number lang ang itype mo!!!")
            else:
                print("ValueError, please input just the number of the temperature.")


def convert_fahrenheit_to_celsius(fahrenheit):                              # Conversion from Fahrenheit to Celsius.
    celsius = (fahrenheit - 32) * (5/9)
    return celsius


def conditions_per_temperature_range(celsius):                              # Suggestions based on temperature.
    if celsius >= 30:
        print("Summer, beach here I come!!")
    elif celsius >= 10:
        print("Go home, it's bed weather!")
    elif celsius >= 0:
        print("Baby it's cold, where are my winter jackets?")
    else:
        print("Hi Olaf. Meet Sven.")

# Main execution flow.
try:
    fahrenheit_val = ask_fahrenheit_input_float()
    celsius_val = convert_fahrenheit_to_celsius(fahrenheit_val)

    print(f"{celsius_val:.2f} degrees celsius")
    conditions_per_temperature_range(celsius_val)

except SystemExit:                                                          # Catches the sys.exit() signal and lets the program end quietly
    pass
```
